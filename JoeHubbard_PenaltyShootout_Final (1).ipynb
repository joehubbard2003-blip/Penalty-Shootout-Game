```python
# PENALTY SHOOTOUT GAME
```

```python
import random

# set up global counters to track results across several shoot-outs
games_played = 0
games_won = 0
games_lost = 0
games_drawn = 0

# Controls whether the player starts another game or not
play_again = "yes"

# Keep running the game if the player chooses to play again
while play_again == "yes":

    print("Welcome to the Penalty Shootout!")

    player_score = 0
    opponent_score = 0

# List of valid positions for the player to shoot, and the goalkeeper to dive
    positions = ["top left", "bottom left", "middle", "top right", "bottom right"]


# Handle the player's penalty, and update their score
    def player_penalty(player_score):
        """Runs the player's penalty and returns their updated score."""
        player_choice = input("Choose from: top left, bottom left, middle, top right, bottom right: ")

# Continue asking for a position until the player correctly types a valid position
        while player_choice not in positions:
            print("Invalid choice. Try again.")
            player_choice = input("Choose from: top left, bottom left, middle, top right, bottom right: ")

        print("You shoot:", player_choice)

        keeper_choice = random.choice(positions)
        print("The goalkeeper dives:", keeper_choice)

        if player_choice == keeper_choice:
            print("SAVED!")
        else:
            print("GOAL!")
            player_score = player_score + 1

        return player_score


# Handle the opponent's penalty, and update their score
    def opponent_penalty(opponent_score):
        """Runs the opponent's penalty and returns their updated score."""
        opponent_choice = random.choice(positions)
        player_keeper_choice = random.choice(positions)

        print("Opponent shoots:", opponent_choice)
        print("Your goalkeeper dives:", player_keeper_choice)

        if opponent_choice == player_keeper_choice:
            print("OPPONENT PENALTY SAVED!")
        else:
            print("OPPONENT SCORES!")
            opponent_score = opponent_score + 1

        return opponent_score

# Play up until five rounds have been completed
    for round_number in range(1, 6):
        print("\nRound", round_number)

        player_score = player_penalty(player_score)

        opponent_score = opponent_penalty(opponent_score)
    
        print("\nRound", round_number)
        print("Your score:", player_score)
        print("Opponent score:", opponent_score)

    games_played = games_played +1

    print("\nFINAL SCORE")
    print("You:", player_score)
    print("Opponent:", opponent_score)

# Compare the final scores and update the game statistics
    if player_score > opponent_score:
        print("YOU WIN!")
        games_won = games_won +1
    elif player_score < opponent_score:
        print("YOU LOSE!")
        games_lost = games_lost +1
    else:
        print("IT'S A DRAW!")
        games_drawn = games_drawn +1

    play_again = input("\nWould you like to play again? yes/no: ").lower()

# Only accept either yes or no when asking if the player wishes to play again
    while play_again != "yes" and play_again != "no":
        print("Invalid choice. Please enter yes or no.")
        play_again = input("Would you like to play again? yes/no: ").lower()


print("\nGAME STATISTICS")
print("Games played:", games_played)
print("Games won:", games_won)
print("Games lost:", games_lost)
print("Games drawn:", games_drawn)

print("Thanks for playing!")
```
